# Motivation

In calculus, in order to define concepts such as **continuity**, **convergence**, **completeness**, and others, we need a notion of **distance**. In finite-dimensional Euclidean spaces, the norm provides this sense of distance. For example, the distance between two vectors $x,y\in \mathbb{R}^n$ is $\|x-y\|$. With this notion of distance, we can define continuity as follows:

A function $f: \mathbb{R}^n \to \mathbb{R}^m$ is continuous if for every $y\in\mathbb{R}^n$ and every $\epsilon > 0$ there exists $\delta > 0$ such that $\|x-y\| < \delta$ implies $\|f(x)-f(y)\| < \epsilon$.

However, there are many norms that can be defined on $\mathbb{R}^n$, and although these notions are largely independent of the particular norm chosen, the definitions still depend on having a norm available.



To address this issue, we can define a notion of distance that is independent of the vector space structure. For example, we can measure the distance between two points on a sphere even though it has no natural underlying vector space structure.
This leads to the more general concept of a **metric space**.

**Definition (Metric space).** A metric space $(X,d)$ consists of a set $X$ and a function $d:X\times X \rightarrow \mathbb{R}$ such that for all $x,y,z\in X$:

1. $d(x,y) \geq 0$.
2. $d(x,y)=0$ if and only if $x=y$.
3. $d(x,y)=d(y,x)$.
4. $d(x,y) \leq d(x,z)+d(z,y)$.

A function $f:X\rightarrow Y$ between metric spaces $(X,d_X)$ and $(Y,d_Y)$ is continuous if for every $y\in X$ and every $\epsilon>0$ there exists $\delta>0$ such that, for every $x\in X$, $d_X(x,y)<\delta$ implies $d_Y(f(x),f(y))<\epsilon$.



### From metric spaces to topology

A metric lets us define open balls $B_d(x,r)=\{y\in X:d(x,y)<r\}$. These open balls, and unions of them, describe which subsets of $X$ are open. Different metrics can give the same collection of open sets, so continuity depends on this collection rather than on the actual distances. This suggests keeping the open sets and discarding the metric altogether.

**Definition (Topology).** A topology on a set $X$ is a collection $\tau$ of subsets of $X$ such that:

1. $\emptyset,X\in\tau$.
2. The intersection of any finite collection of sets in $\tau$ belongs to $\tau$.
3. The union of any collection of sets in $\tau$ belongs to $\tau$.

Every metric gives rise to a topology, but some topologies do not come from any metric.

The pair $(X,\tau)$ is called a **topological space**. This structure lets us define continuity without measuring distances: 

A function $f:X\rightarrow Y$ between topological spaces is continuous if for every open set in $U \in Y$, $f^{-1}(U)$ is open in $X$. 

The following are some important concepts in topology:

**Definition (Closed set):** $U\subset M$ is **closed** if its complement $\overline{U}$ is open.

**Definition (Continuity and homeomorphism):** Let $(M,\tau)$ and $(N,\sigma)$ be topological spaces. A function $f:M\rightarrow N$ is **continuous** if whenever $U\in\sigma$, $f^{-1}(U)\in\tau$.Furthermore, if $f$ is a bijection with continuous inverse, then it is a **homeomorphism**.

**Definition (Open cover):** A collection of open sets $V_a$ is an open cover of the set $X$ if $X \subset  \bigcup_{a}^{n} V_a$. 
A subcover is a subcollection of $V_a$  that still covers $X$.

**Definition (Compactness):** A topological space is **compact** if every open cover contains a finite subcover.

**Definition (Hausdorff space):** A topological space is **Hausdorff** (or $T_2$) if for every pair of distinct points $x,y$, there exist open sets $U\ni x$ and $V\ni y$ such that $U\cap V=\emptyset$.
