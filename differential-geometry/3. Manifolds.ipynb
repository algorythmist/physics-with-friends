# Manifolds


### Manifolds

**Definition (Locally Euclidean).** A topological space $M$ is called **locally Euclidean** if there exists a nonnegative integer $n$ such that every point in $M$ has a neighborhood homeomorphic to $\mathbb{R}^n$.

**Definition (Topological manifold).** A **topological manifold** is a Hausdorff, second-countable, locally Euclidean space.

**Definition (Coordinate map and coordinate system).** If $M$ is locally Euclidean and $U\subset M$ is open, a homeomorphism $\phi:U\to\mathbb{R}^n$ is called a **coordinate map**. The pair $(U,\phi)$ is a **coordinate system**, and the functions $\phi^i(p)$ are called **coordinates**. A collection of coordinate systems whose domains cover $M$ is called an **atlas**.

**Definition (Differentiable structure).** A **differentiable structure** $\mathcal{F}$ on a locally Euclidean space $M$ is a collection of coordinate systems $(U_a,\phi_a)$ satisfying:

1. $\bigcup_a U_a = M$.
2. $\phi_a\circ\phi_b^{-1}$ is differentiable for all $a,b$.
3. $\mathcal{F}$ is maximal: if $\phi\circ\phi_a^{-1}$ and $\phi_a\circ\phi^{-1}$ are differentiable, then $(U,\phi)\in\mathcal{F}$.

**Definition (Manifold).** A **manifold** is a second-countable, Hausdorff, locally Euclidean space equipped with a differentiable structure.


### Projective Space

**Definition (Real and complex projective spaces).** A **real projective space** $\mathbb{R}P^n$ is the space of unoriented lines through the origin in $\mathbb{R}^{n+1}$. Similarly, **complex projective space** $\mathbb{C}P^n$ is the space of complex lines through the origin in $\mathbb{C}^{n+1}$.

The following discussion works for both real and complex spaces, so let $\mathbb{F}$ denote either $\mathbb{R}$ or $\mathbb{C}$. (Question: what about $\mathbb{Q}$, or the quaternions?)

To any point $p=(z_0,z_1,\ldots,z_n)\in\mathbb{F}^{n+1}$, we associate the equivalence class of all nonzero scalar multiples of $p$, namely $\lambda(z_0,z_1,\ldots,z_n)$ for $\lambda\neq 0$. We call this equivalence class $[z_0,z_1,\ldots,z_n]$ the **homogeneous coordinates** of the line through $p$.

If $z_i\neq 0$ for some $i$, define

$$\phi_i(p)=\left(\frac{z_0}{z_i},\frac{z_1}{z_i},\ldots,\widehat{\frac{z_i}{z_i}},\ldots,\frac{z_n}{z_i}\right).$$

This chart removes the $i$th coordinate after scaling so that the $i$th component becomes $1$.
