# Overview

## Vectors and Forms

At each point $p \in \mathbb{R}^n$, the tangent space $V = T_p\mathbb{R}^n$ is a vector space. In local coordinates,

$$
\mathbf{v} = v^i\boldsymbol{\partial}_i
$$

Vectors act on functions at $p$ by differentiation:

$$
\mathbf{v}(f) = v^i\boldsymbol{\partial}_i f
$$

A 1-form is an element of the cotangent space $T_p^*\mathbb{R}^n$, dual to $T_p\mathbb{R}^n$:

$$
\alpha(\mathbf{v}) = a_i v^i
$$

For a function $f$, define its differential $df \in T_p^*\mathbb{R}^n$ by

$$
df(\mathbf{v}) := \mathbf{v}(f)
$$

The coordinate 1-forms $\{dx^i\}$ form a basis, so $df = (\partial_i f)\,dx^i$.

## Tensor Products

For two one-forms,

$$
(\alpha \otimes \beta)(\mathbf{v},\mathbf{w}) := \alpha(\mathbf{v})\beta(\mathbf{w})
$$

The tensors $dx^i \otimes dx^j$ form a basis for covariant rank-2 tensors.

Similarly, for vectors $\mathbf{v},\mathbf{w} \in V$, their tensor product $\mathbf{v} \otimes \mathbf{w} \in V \otimes V$ has basis $\boldsymbol{\partial}_i \otimes \boldsymbol{\partial}_j$.

Tensor products combine associatively to form higher-order and mixed tensors. For example,

$$
A = A^{ij}_k \boldsymbol{\partial}_i \otimes \boldsymbol{\partial}_j \otimes dx^k
$$

## Exterior Product

For 1-forms, we define

$$
\alpha \wedge \beta := \alpha \otimes \beta - \beta \otimes \alpha
$$

A 2-form can be written as $\omega = \sum_{i < j} \omega_{ij}\,dx^i \wedge dx^j$.

The exterior product extends associatively to higher-degree forms. In particular, for any 1-form $\alpha$, $\alpha \wedge \alpha = 0$, so nonzero forms have degree at most $n$ on $\mathbb{R}^n$.

For a $p$-form $\alpha$ and a $q$-form $\beta$,

$$
\alpha \wedge \beta = (-1)^{pq}\,\beta \wedge \alpha
$$

The exterior derivative of a $p$-form is defined by

$$
d(a\,dx^{i_1}\wedge\cdots\wedge dx^{i_p}) := da\wedge dx^{i_1}\wedge\cdots\wedge dx^{i_p} = \frac{\partial a}{\partial x^r}\,dx^r\wedge dx^{i_1}\wedge\cdots\wedge dx^{i_p}.
$$

## Metrics

A metric is an inner product $\langle\cdot,\cdot\rangle$ on $V$. Its components in local coordinates are

$$
g_{ij} = \langle\boldsymbol{\partial}_i,\boldsymbol{\partial}_j\rangle.
$$

$$
\langle\mathbf{v},\mathbf{w}\rangle = g_{ij}v^i w^j
$$

The associated metric tensor is

$$
ds^2 = g_{ij}\,dx^i \otimes dx^j.
$$

It satisfies

$$
ds^2(\mathbf{v},\mathbf{w}) = \langle\mathbf{v},\mathbf{w}\rangle.
$$

The metric identifies each vector $\mathbf{v} \in V$ with a covector $\mathbf{v}^{\flat} \in V^*$:

$$
\mathbf{v}^{\flat}(\mathbf{w}) := \langle\mathbf{v},\mathbf{w}\rangle
$$

In coordinates, $v_i = g_{ij}v^j$.

The gradient is defined by $df(\mathbf{w}) = \langle\boldsymbol{\nabla}f,\mathbf{w}\rangle$.

### The volume form

On oriented $\mathbb{R}^n$, let $g_{ij}$ be the components of a Riemannian metric and let $g = \det(g_{ij})$. The volume form is

$$
\mathrm{V^{(n)}} := \sqrt{g}\,dx^1\wedge\cdots\wedge dx^n.
$$

## Pushing and Pulling

Let $F: \mathbb{R}^k \to \mathbb{R}^n$ have coordinates $u=(u^1,\ldots,u^k)$ in the domain and $x=(x^1,\ldots,x^n)$ in the codomain, with $x=F(u)$.

The pushforward sends a vector $\mathbf{v}$ at $u$ to a vector $F_*\mathbf{v}$ at $F(u)$, with components

$$
[F_*\mathbf{v}]^i = \frac{\partial x^i}{\partial u^j}v^j.
$$

In particular, $F_*(\boldsymbol{\partial}_j) = \frac{\partial x^i}{\partial u^j}\boldsymbol{\partial}_i$.

For an $m$-form $\alpha$ on $\mathbb{R}^n$, the pullback $F^*\alpha$ on $\mathbb{R}^k$ is defined by

$$
(F^*\alpha)(\mathbf{v}_1,\ldots,\mathbf{v}_m) := \alpha(F_*\mathbf{v}_1,\ldots,F_*\mathbf{v}_m).
$$

For a 0-form $f$, $(F^*f)(u)=f(F(u))$. For coordinate 1-forms, $F^*dx^i = \frac{\partial x^i}{\partial u^j}\,du^j$.

**Properties**

- $F^*(\alpha \wedge \beta) = F^*\alpha \wedge F^*\beta$
- $F^*(d\alpha) = d(F^*\alpha)$

## Integration

For a 1-form $f(x)\,dx$ on an interval $[a,b]\subset\mathbb{R}$,

$$\int_{[a,b]} f(x)\,dx = \int_a^b f(x)\,dx.$$

Let $\gamma:[a,b]\to\mathbb{R}^n$ parametrize a curve $C$, and let $\alpha$ be a 1-form on $\mathbb{R}^n$. Its line integral is

$$
\int_C \alpha := \int_{[a,b]} \gamma^*\alpha.
$$

More generally, if $V$ is an oriented $p$-dimensional manifold with boundary and $\beta$ is a $(p-1)$-form on $V$, Stokes' theorem states

$$
\int_V d\beta = \int_{\partial V} \beta.
$$

## Interior Product

Let $\beta$ be a $p$-form and $\mathbf{v}$ a vector. The interior product $i_{\mathbf{v}}\beta$ is the $(p-1)$-form defined by

$$
(i_{\mathbf{v}}\beta)(\mathbf{A}_2,\ldots,\mathbf{A}_p) := \beta(\mathbf{v},\mathbf{A}_2,\ldots,\mathbf{A}_p).
$$

In particular, for a 1-form,

$$
i_{\mathbf{v}}\beta = \beta(\mathbf{v}) = \beta_i v^i.
$$

The interior product satisfies

- $i_{\mathbf{v}}(f\alpha) = f\,i_{\mathbf{v}}\alpha$ for any function $f$.

- $i_{\mathbf{v}}(\alpha\wedge\beta) = (i_{\mathbf{v}}\alpha)\wedge\beta + (-1)^p\alpha\wedge(i_{\mathbf{v}}\beta)$, where $\alpha$ is a $p$-form.

We can define the divergence of a vector field in terms of the volume form:

$$
V^n(\nabla \cdot V) := d(i_v(V^n)) = d(\sqrt{g}dx^1\ldots dx^n) = \frac{\partial(\sqrt{g}v^i)}{\partial x^i}dx^1\ldots dx^n
$$

so that

$$
\nabla \cdot V := \frac{1}{\sqrt{g}} \frac{\partial(\sqrt{g}v^i)}{\partial x^i}
$$
