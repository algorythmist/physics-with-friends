### Problem 1.1(1)

In cylindrical coordinates, $z = \pm \sqrt{r^2 - c}$

Run the plot cell to display the interactive surfaces. Drag the graph to rotate; use the scroll wheel to zoom.

```python
import numpy as np
import plotly.graph_objects as go
from plotly.colors import qualitative

r = np.linspace(0, 3, 100)
theta = np.linspace(0, 2 * np.pi, 120)
R, Theta = np.meshgrid(r, theta)

X = R * np.cos(Theta)
Y = R * np.sin(Theta)
Z_sqrt_r2_minus_1 = np.full_like(R, np.nan)
mZ_sqrt_r2_minus_1 = np.full_like(R, np.nan)
valid_domain = R >= 1
Z_sqrt_r2_minus_1[valid_domain] = np.sqrt(R[valid_domain]**2 - 1)
mZ_sqrt_r2_minus_1[valid_domain] = -np.sqrt(R[valid_domain]**2 - 1)

surfaces = [
    (R, "z = r"),
    (-R, "z = -r"),
    (np.sqrt(R**2 + 1), "z = sqrt(r² + 1)"),
    (-np.sqrt(R**2 + 1), "z = -sqrt(r² + 1)"),
    (Z_sqrt_r2_minus_1, "z = sqrt(r^2 - 1), r >= 1"),
    (mZ_sqrt_r2_minus_1, "z = -sqrt(r^2 - 1), r >= 1"),
]

fig = go.Figure()

colors = ["teal", "coral", "white"]
for index, (Z, name) in enumerate(surfaces):
    c = int(index/2)
    color = colors[c]
    fig.add_trace(go.Surface(
        x=X, y=Y, z=Z, name=name, opacity=0.5, showscale=False,
        colorscale=[[0, color], [1, color]],
        hovertemplate="x=%{x:.2f}<br>y=%{y:.2f}<br>z=%{z:.2f}<extra>" + name + "</extra>",
    ))

fig.update_layout(
    title="Sub manifolds",
    scene={
          "camera": {
            "eye": {"x": 1.5, "y": 1.5, "z": 0.8},
            "up": {"x": 0, "y": 0, "z": 1},
        },
        "aspectmode": "manual",
        "aspectratio": {"x": 1, "y": 1, "z": 1.3},
        "xaxis_title": "x",
        "yaxis_title": "y",
        "zaxis_title": "z",
        #"aspectmode": "data",
    },
    margin={"l": 0, "r": 0, "t": 45, "b": 0},
    legend={"x": 0.02, "y": 0.98},
  
)
fig.show()
```

- $c = 0$ is a disconnected manifold (2 cones) excluding the origin
- $c = 1$ is  a disconnected manifold excluding the unit circle
- $c = -1$ is a disconnected manifold including two paraboloids

### Problem 1.1(4)

The $x$ component of $\nabla F \times \nabla G$ is 
$
\partial_y F \partial_z G - \partial_z F \partial_y G = det \begin{bmatrix} \partial_y F & \partial_z F\\  \partial_y G &  \partial_z G 
\end{bmatrix}
$ 

Differentiating F = 0 and G = 0 with respect to x:

$$
\partial_x F + \partial_y F \frac{dy}{dx}  + \partial_z F \frac{dz}{dx} =0 
$$

$$
\partial_x G + \partial_y G \frac{dy}{dx}  + \partial_z G \frac{dz}{dx} = 0
$$


$$
\begin{bmatrix} \partial_y F & \partial_z F \\
\partial_y G & \partial_z G \end{bmatrix}
\begin{bmatrix} \frac{dy}{dx} \\ \frac{dz}{dx} \end{bmatrix} =
-\begin{bmatrix} \partial_x F \\ \partial_x G \end{bmatrix}
$$

This system of equations can be solved locally, since the matrix is invertible, to give a parametrization for the curve.
